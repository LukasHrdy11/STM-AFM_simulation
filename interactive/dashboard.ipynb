## Náměty, co studentům ukázat

**1. Co dělá regulátor (demystifikace).** V STM panelu klikni na „Načíst
kroky" a posouvej „Krok". Vypíše se každý mezikrok smyčky: změřený proud →
log-chyba `e = ln(I/I_set)` → příkaz regulátoru → nová poloha hrotu.
Regulátor neví, kde povrch je — ví jen, jestli je proud větší nebo menší,
než má být.

**2. Zpětná vazba zapnutá vs. vypnutá.** Odškrtni „zpětná vazba zapnutá".
Hrot přestane uhýbat a topografie se promítne rovnou do proudu (režim
konstantní výšky) — proud přes hranu vyskočí exponenciálně. Pak stáhni
posuvník „z_hrot" (zpřístupní se právě v tomhle režimu) pod výšku hrany:
hrot do ní narazí, protože ho nemá co zvednout. To je přesně ten rozdíl
mezi režimem konstantního proudu a konstantní výšky.

Pro srovnání v jednom grafu: nejdřív si se zapnutou smyčkou klikni na
„Zamknout křivku", pak smyčku vypni — zamknutá dráha hrotu zůstane
v grafu šedě na pozadí.

**3. Kdy je šum neúnosný.** Posuvník „Míra šumu" je násobek skutečně
změřeného šumu. Při 1× je šum jednotky pm proti rezervě stovek pm.
Postupně zvyšuj a sleduj, kdy regulátor začne ztrácet stopu povrchu a kdy
rezerva do nárazu spadne k nule.

**4. P vs. I vs. PI.** Nastav stejný povrch a šum, klikni „Zamknout
křivku", přepni regulátor a porovnej v jednom grafu. Pozor: samotná
P-složka má jiný efekt než přeškálování zesílení jen při `T_sys > 0`.

**5. Co je Δf.** V AFM panelu posouvej Δf u rezonanční křivky. Δf není
měřená veličina — je to posun celé rezonance, ze kterého se teprve
odečítá.

**6. Ztráta stability u AFM.** Přepni citlivost na „local" a povrch na
`atoms`. Jakmile se hrot dostane za minimum Δf(d), citlivost změní
znaménko a smyčka reguluje opačně. Je to očekávaný jev, ne chyba
simulace.
